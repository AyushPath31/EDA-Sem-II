```python
#Declaration of NumPy

import numpy as np

a = np.array([10,20,30,40])
b = np.array([[1,2,3],[4,5,6]])
print(a)
print(b)
```

```python
#Dimension of Array

print(a.ndim)
print(b.ndim)
```

```python
#Shape of Array

print(a.shape)
print(b.shape)
```

```python
#Datatype

print(a.dtype)
print(b.dtype)
```

```python
#In-built functions using numpy

#Initialise zeroes matrix
np.zeros((2,3))
```

```python
#Initialise ones matrix
np.ones((3,3))
```

```python
#Initialise identity matrix

np.eye((3))
```

```python
#Initialise Array Range

np.arange(1,10,2)
```

```python
#Initialise Linespacing
np.linspace(0,1,5)
```

```python
c = a * 2
d = a + 5
print(c)
print(d)
```

```python
a + 5
```

```python
#Average, Median
print(np.mean(a))
print(np.median(a))
```

```python
# Max and Min value
print("Maximum Value from 'a' array: ",np.max(a))
print("Minimum Value from 'a' array: ",np.min(a))
```

```python
#Sum Function
np.sum(a)
```
