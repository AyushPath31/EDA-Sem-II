```python
#Dataset-1

import pandas as pd

data = {
    "Roll No": [101, 102, 103, 104, 105],
    "Name": ["Asha", "Rohan", "Meera", "Karan", "Anil"],
    "Age": [20, "-", 21, 20, "22"],
    "Marks": [85, 90, "-", 88, 76],
    "Department": ["CSE", "IT", "CSE", "cse", "E&TC"],
    "Admission_Date": ["01-06-2023","03-06-2023","05-06-2023","07-06-2023","10/06/23"]
}

df = pd.DataFrame(data)

# Save as CSV
df.to_csv("student_data.csv", index=False)

df
```

```python
# Step 2: Replace '-' with NaN to mark missing values

df.replace("-", np.nan, inplace=True)

# Convert columns to numeric where needed
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

'''
When converting data (strings --> datetime), if some values are invalid or
cannot be converted, instead of throwing an error, it will: replace invalid values
with NaT (Not a Time).
'''
df
```

```python
# Step 3: Check Missing Values
print("Missing Values:")
print(df.isnull().sum())
```

```python
import warnings
warnings.filterwarnings("ignore")


# Fill missing age with mean age
df["Age"].fillna(df["Age"].mean(), inplace=True)

#Fill missing marks with median
df["Marks"].fillna(df["Marks"].median(), inplace=True)

df
```

```python
print(df.isnull().sum())
```

```python
# Step 5: Fix inconsistent text formatting

df["Department"] = df["Department"].str.upper()
df
```

```python
# Step 6: Convert Admission_Date to standard datetime format

df["Admission_Date"] = pd.to_datetime(df["Admission_Date"], errors='coerce')
df

# If a value cannot be converted to valid date, pandas will replace it with NaT
```

```python
# Step 7: Save cleaned dataset

df.to_csv("cleaned_studentdata.csv",index=False)
print("Cleaned CSV file saved as cleaned_studentdata.csv")
```

Wokring on Cars93 Dataset

```python
# Working with missing values DataSET 2

import pandas as pd
import numpy as np

df=pd.read_csv("Cars93.csv")
df.info()
```

```python
df.isna().sum() #Attributes are displayed
```

Imputing the Missing Value

```python
# Imputing the Missing Value
# 1. Luggage Room

df['Luggage.room'] = df['Luggage.room'].fillna(value=df['Luggage.room'].mean())
df.isna().sum()
```

```python
# 2. Rear Seat Room
df['Rear.seat.room'] = df['Rear.seat.room'].fillna(value=df['Rear.seat.room'].mean())
df.isna().sum()
```

```python
# 3. Air Bags
df['AirBags'] = df['AirBags'].fillna(value=df['AirBags'].mode().values[0])
df.isna().sum()
```

```python
# Save cleaned data
df.to_csv("cleaned_cars93.csv",index=False)
print("Cleaned CSV file saved as cleaned_cars93.csv")
```
