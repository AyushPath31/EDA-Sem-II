```python
'''
1. Nominal Data
Categories without any order
Ex - Gender, Blood Grp, Department


2. Ordinal Data
Categories with meaningful order
Ex - Ratings (Poor, Avg, Good), Grades (A,B,C)

Common Categorical Data Analysis Methods
Frequency Distribution
Percentage Distribution
Value Counts

'''
```

```python
import pandas as pd

data = {
    'Order_ID':['01','02','03','04','05','06','07','08','09','010'],
    'Category':['Electronics','Clothing','Electronics','Grocery','Clothing',
                'Electronics','Grocery','Clothing','Electronics','Grocery'],
    'Payment_Method':['UPI','Card','UPI','Cash','Card',
                      'UPI','Cash','Card','UPI','Cash'],
    'Delivery_Type':['Express','Standard','Express','Standard','Express',
                     'Standard','Express','Standard','Express','Standard'],
    'Customer_Type':['New','Returning','Returning','New','Returning',
                      'New','Returning','New','Returning','New']
}
df = pd.DataFrame(data)
df
```

```python
# 1. Frequency Count
# Count How many Orders in Each Category

print(df['Category'].value_counts())
```

```python
# 2. Unique Values
# Display all unique categories

print(df['Category'].unique())
```

```python
# 3. Number of Unique Values
print("Number of Unique Categories",end=" - ")
print(df['Category'].nunique())
```

```python
# 4. Frequency of Payment Methods

print(df['Payment_Method'].value_counts())
```

```python
# 5. Cross Tabulation
# Category VS Payment Methods
print(pd.crosstab(df['Category'],df['Payment_Method']))
```

```python
# 6. Cross Tabulation ( Delivery VS Customer Type)
print(pd.crosstab(df['Delivery_Type'],df['Customer_Type']))
```

```python
# 7. Percentage Distribution
print(df['Delivery_Type'].value_counts(normalize=True)*100)
```

```python
# 8. Filtering Data
# Example: Orders from Electronics Category

electronics_orders = df[df['Category']=='Electronics']
print(electronics_orders,end="\n\n")

clothes = df[df['Category']=='Clothing']
print(clothes)
```

```python
# 9 . Grouping
# Orders by category and payment method

print(df.groupby('Category')['Payment_Method'].value_counts())
```

```python
# 10. Sorting Data
sorted_df = df.sort_values(by='Category')
print(sorted_df)
```

Working on New Data Set

```python
import pandas as pd
import numpy as np
df = pd.read_csv('Expt11.csv')
df.info()
```

```python
print(df.head(),end="\n\n\n")
print(df.tail())
```

```python
# 1. Count Students in Each Grade
# Find How many students obtained each grade

print(df['Grade'].value_counts().sort_index())
# By default value_counts() displays data according to frequency in descending order.
# Use sort_index() to display data according to index
```

```python
# 2. Percentage Distribution of Grades

output = df['Grade'].value_counts(normalize=True)*100
output.sort_index(inplace=True)
print(output)
```

```python
# 3. Number of Students in each department

print(df['Department'].value_counts())
```

```python
# 4. Gender Frequency Distribution
print(df['Gender'].value_counts())
```

```python
# 5. Gender VS Grade Analysis
print(pd.crosstab(df['Gender'],df['Grade']))
```

```python
# 6. Department vs Gender
print(pd.crosstab(df['Department'],df['Gender']))
```

```python
# 7. Department vs Grade Percentage

print((pd.crosstab(df['Department'], df['Grade'], normalize=True) * 100).round(2))
```

```python
# 8. Grouping Data (Departmentwise)
print(df.groupby('Department')['Gender'].value_counts())
```
