```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days' : ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Study_Hours' : [2,3,2.5,4,3.5],
    'Marks' : [50,60,55,70,65],
    'Attendance' : [60,70,65,80,75],
    'Sleep_Hours' : [6,7,6,8,7],
    'Assignments_Completed' : [1,2,1,3,2]
}

df = pd.DataFrame(data)
df
```

```python
# 1. Line Chart (Trend)
# plt.figure()
plt.plot(df['Days'],df['Study_Hours'],marker='o')

plt.title("Study Hours Trend")
plt.xlabel("Days")
plt.ylabel("Study Hours")
plt.show()
```

```python
# Advance Line Chart

plt.figure(figsize=(7,4))

plt.plot(df['Days'], df['Study_Hours'], marker='^', color='green',label='Study Hours')
plt.plot(df['Days'], df['Marks'], marker='s', color='red',label='Marks')

plt.title("Study Hours and Marks Trend")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()
plt.show()
```

### Common Marker Types

'x' - Cross

```python
# 2. Bar Chart (Comparison)

plt.bar(df['Days'],df['Marks'], color='skyblue')

plt.title("Marks Comparison")
plt.xlabel("Days")
plt.ylabel("Marks")
plt.show()
```

```python
# Advanced Bar Chart

bars = plt.bar(df['Days'],df['Marks'], color='maroon')

for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2, y, str(y), ha='center', va='bottom')

plt.title("Marks Comparison")
plt.xlabel("Days")
plt.ylabel("Marks")

plt.grid(axis='y') # Adds Horizontal Grid for readability

plt.show()
```

```python
# 3. Histogram (Distribution)

# Shows how values are spread

data = {
    'Marks' : [50,60,55,70,65,75,80,85,90,45,68,72,77,82,88]
}

df = pd.DataFrame(data)

plt.hist(df['Marks'], bins=3, edgecolor='red')
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4. Scatter Plot (Relationship)
import matplotlib.pyplot as plt

plt.scatter(df['Study_Hours'],df['Marks'])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Conditional Scatter Plot
# Example Category

df['Result'] = ['Fail','Pass','Fail','Pass','Pass']

colors = ['red' if r=='Fail' else 'green' for r in df['Result']]

plt.scatter(df['Study_Hours'],df['Marks'], c=colors)

plt.title("Study Hours vs Marks [Result-Wise]")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

### Visualisations using Seaborn

```python
# Dataset 2

data = {
    'Days' : ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed'],
    'Region' :['North','South','East','West','North','South','East','West','North','South'],
    'Sales' : [200,220,250,270,300,350,400,280,260,290],
    'Profit' : [20,25,30,35,50,60,80,40,38,45],
    'Customers' : [30,35,40,45,50,65,80,55,48,52],
    'Category' : ['Food','Clothing','Electronics','Food','Clothing','Electronics','Food','Clothing','Electronics','Food']
}

df = pd.DataFrame(data)
df
```

```python
# Line Plot

sns.lineplot(x='Days',y='Sales',data=df)
plt.title("Seaborn Line Plot")
plt.show()
```

```python
# Histogram

sns.histplot(df['Sales'],bins=5)
plt.title("Seaborn Histogram")
plt.show()
```

```python
# Scatter Plot

sns.scatterplot(x='Sales',y='Profit',data=df)
plt.title("Seaborn Scatter Plot")
plt.show()
```
