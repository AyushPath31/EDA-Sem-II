```python
#Dataset

import pandas as pd
import numpy as np

data = {
    'Product' : ['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
    'Price' : [55000,20000,30000,2000,8000,45000],
    'Units_Sold':[25,60,35,80,50,15],
    'Discount' : [10,15,12,20,18,8]
}

df = pd.DataFrame(data)

print("Original Dataset")
print(df)
```

```python
#1. Min-Max Normalization
# for Price

df['Price_MinMax'] = (df['Price'] - df['Price'].min()) / (df['Price'].max() - df['Price'].min())

print("\nMin-Max Normalized Price:")
print(df[['Product','Price', 'Price_MinMax']])
```

```python
#1. Min-Max Normalization
# for Discount

df['Discount_MinMax'] = (df['Discount'] - df['Discount'].min()) / (df['Discount'].max() - df['Discount'].min())
print("\nMin-Max Normalized Discount:")
print(df[['Product','Discount', 'Discount_MinMax']])
```

```python
#2. Z-Score Normalization
df['Units_Zscore'] = (df['Units_Sold'] - df['Units_Sold'].mean()) / df['Units_Sold'].std()

print("\nZ-Score Normalized Units Sold:")
print(df[['Product','Units_Sold', 'Units_Zscore']])
```

```python
#3. Decimal Scaling - For Price
df['Price_Decimal'] = df['Price'] / 100000

print("\nDecimal Scaled Price:")
print(df[['Product','Price', 'Price_Decimal']])
```

```python
#3. Decimal Scaling - For Discount
df['Discount_Decimal'] = df['Discount'] / 100

print("\nDecimal Scaled Discount:")
print(df[['Product','Discount', 'Discount_Decimal']])
```

```python
# Normalize Multiple Columns

cols = ['Price', 'Units_Sold', 'Discount']
df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())

print("\nNormalized Dataset")
print(df_norm)
```

```python
data = {
        'Order_ID':['101','102','103','104','105','106'],
        'Customer_Gender' : ['Male','Female','Female','Male','Female','Male'],
        'Payment_Method' : ['UPI','Credit Card','Debit Card','COD','UPI','Debit Card'],
        'Product_Category' : ['Electronics','Clothing','Home','Electronics','Beauty','Elctronics'],
        'City' : ['Pune','Mumbai','Delhi','Bangalore','Pune','Hyderabad'],
        'Order_Value' : [4500,2300,3200,5200,1800,4100]
}
df = pd.DataFrame(data)
df
```

```python
# Method 1 : Label Encoding

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Gender_Label'] = le.fit_transform(df['Customer_Gender'])
df
```

```python
# Method 2 : One-Hot Encoding

df_encoded = pd.get_dummies(df, columns=['Payment_Method'])

print(df_encoded)
```

```python
# Method 3 : Encoding Multiple Columns
df_multi = pd.get_dummies(df, columns = ['Product_Category','City'])
print(df_multi.head())
```

```python
# Method 4 : Dummy Encoding (Drop First)
df_dummy = pd.get_dummies(df, columns = ['Payment_Method'], drop_first = True)
print(df_dummy)
```

```python
import pandas as pd
df = pd.read_csv('/content/amazon_products_dataset_Expt-14.csv')

df.info()
```

```python
# Data Normalization - Multiple Columns [MinMax Method]

# Normalize Multiple Columns

cols = ['Price','Rating','Reviews', 'Units_Sold']
df_norm = (df[cols] - df[cols].min()) / (df[cols].max() - df[cols].min())

print("\nNormalized Dataset")
print(df_norm)
```

```python
# Data Normalization - Z-Score Normalization

cols_2 = ['Price','Rating','Reviews', 'Units_Sold']
df_norm2 = (df[cols_2] - df[cols_2].mean()) / df[cols_2].std()

print("\nZ-Score Normalized Units Sold:")
print(df_norm2)
```

```python
# Data Normalization - Decimal Scaling

df_price = df['Price'] / 100000
df_rating = df['Rating'] / 10
df_reviews = df['Reviews'] / 10000
df_units = df['Units_Sold'] / 1000

df_norm = pd.concat([df_price,df_rating,df_reviews,df_units], axis = 1)
print("\nDecimal Scaled Values:")
print(df_norm)
```
