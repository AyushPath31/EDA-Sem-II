```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
np.random.seed(0)

df = pd.DataFrame ({
    'Category' : ['A', 'B', 'C', 'D', 'E'],
    'Value' : [50,70,30,90,60],
    'Sales' : np.random.randint(100,500,5),
    'Profit' : np.random.randint(10,100,5)

})

print(df.head())
df.info()
```

```python
# 1. Area Plot

plt.figure()
plt.fill_between(df['Category'],df['Value'], color='blue',alpha=0.55, edgecolor='red')
plt.title('Area Plot')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

```python
# 2. Pie Chart

plt.figure()
plt.pie(df['Value'],labels=df['Category'],autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

```python
# 3. Donut Chart

plt.figure()
plt.pie(df['Value'],labels=df['Category'],autopct='%1.2f%%')
centre_circle = plt.Circle((0,0),0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Donut Chart")
plt.show()
```

```python
#4. Boxplot (outliers detection)

plt.figure()
sns.boxplot(x=df['Value'])
plt.title('Boxplot of Sales')
plt.show()
```

```python
#5. headtop(correlation)

plt.figure()
corr=df[['Sales','Profit']].corr()
sns.heatmap(corr,annot=True,cmap='cool')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#Bubble Plot

import matplotlib.pyplot as plt
plt.scatter(df['Sales'],df['Profit'],s=df['Value']*15,edgecolors='pink',linewidths=4)    #size of bubble
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Bubble Plot')
plt.show()
```

```python
# Seaborn Bubble Plot

plt.figure(figsize=(6,4))
sns.scatterplot(x='Sales',y='Profit',size='Value',hue='Value',data=df,sizes=(50,300),palette='viridis')   #hue gives color and size gives bubble size, sizes gives min max
plt.title("Bubble plot using seaborn")
plt.show()
```
