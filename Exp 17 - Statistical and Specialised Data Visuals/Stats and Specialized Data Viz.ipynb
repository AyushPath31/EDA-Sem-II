```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(0)
df=pd.DataFrame({
    'Category':['A','B','C','D','E'],
    'Values':[50,70,30,90,60],
    'Sales':np.random.randint(100,500,5),
    'Profit':np.random.randint(10,100,5)
})
print(df.head())
```

```python
#1. Area Plot
plt.figure()
plt.fill_between(df['Category'],df['Values'],color='blue',alpha=0.5)
plt.title('Area Plot')
plt.xlabel('Category')
plt.ylabel('Values')
plt.show()
```

```python
#2. PIE CHART
plt.figure()
plt.pie(df['Values'],labels=df['Category'],autopct='%1.1f%%')
plt.title('PIE CHART')
plt.show()
```

```python
#3. DONUT CHART
plt.figure()
plt.pie(df['Values'],labels=df['Category'],autopct='%1.1f%%')
centre_circle=plt.Circle((0,0),0.70,fc='white')
fig=plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title('DONUT CHART')
plt.show()
```

```python
#4. Boxplot (outliers detection)
plt.figure()
sns.boxplot(x=df['Values'])
plt.title('Boxplot of Sales')
plt.show()
```

```python
#5. headtop(correlation)
plt.figure()
corr=df[['Sales','Profit']].corr()
sns.heatmap(corr,annot=True,cmap='cool')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#bubble plot
import matplotlib.pyplot as plt
plt.scatter(df['Sales'],df['Profit'],s=df['Values']*15,edgecolors='pink',linewidths=4)    #size of bubble
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.title('Bubble Plot')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Sales',y='Profit',size='Values',hue='Values',data=df,sizes=(50,300),palette='viridis')   #hue gives color and size gives bubble size, sizes gives min max
plt.title("Bubble plot using seaborn")
plt.show()
```

```python
#2. Second Dataset
np.random.seed(7)
n = 100
df = pd.DataFrame({
    'Customer_ID': range(1, n+1),
    'Age': np.random.randint(20, 60, n),
    'Income': np.random.randint(20000, 100000, n),
    'Loan_Amount': np.random.randint(5000, 50000, n),
    'Credit_Score': np.random.randint(300, 900, n)
})

df['Loan_Status'] = np.where(df['Credit_Score'] > 600, 'Approved', 'Rejected')

print(df.head())
```

```python
#Boxplot (Outlier Detection – Loan Amount)
import matplotlib.pyplot as plt

plt.boxplot(df['Loan_Amount'])

plt.title("Loan Amount Distribution (Boxplot)")
plt.ylabel("Loan Amount")

plt.show()
```

```python
#3. Histogram (Credit Score Distribution)
plt.hist(df['Credit_Score'], bins=6, color='red', edgecolor='black')

plt.title("Credit Score Distribution")
plt.xlabel("Credit Score")
plt.ylabel("Frequency")

plt.show()
```

```python
#Bubble Plot (Advanced Scatter)
plt.scatter(df['Income'], df['Loan_Amount'],
            s=df['Credit_Score']/2,
            c=df['Credit_Score'],
            cmap='viridis',
            alpha=0.6)

plt.colorbar(label='Credit Score')

plt.title("Bubble Plot (Income vs Loan)")
plt.xlabel("Income")
plt.ylabel("Loan Amount")

plt.show()
```

```python
#6. Correlation Heatmap (Very Important)
import seaborn as sns

corr = df[['Age','Income','Loan_Amount','Credit_Score']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title("Correlation Matrix")

plt.show()
```
