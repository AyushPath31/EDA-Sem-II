# Sets
Sets are used to store multiple items in a single variable  
Sets items are ordered, unchangeable, and do not allow duplicate values. Unordered means that items in a set do not have a defined order. Set items can appear in a different order everytime you use them.

```python
# Creation of set
myset = {"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin"}
print(myset)
```

```python
# Duplicates not allowed
myset = {"Ferrari","Haas","McLaren","Ferrari","Ferrari"}
print(myset)
```

```python
# The values True and 1 are considered the same values in sets
myset = {"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin",True, 1, 2, 3, 4}
print(myset)
```

```python
# A set with strings, integers and boolean values:
setA = {"xyz",99,True,False,44,"wild west",78.992}
setA
```

```python
# To check membership
myset = {"Ferrari","Haas","McLaren","Ferrari","Ferrari"}
print("BMW" in myset)
print("Haas" in myset)
```

```python
# Once a set is created, you cannot replace its items, but you can add new items.
myset = {"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin"}
myset.add("BMW")
print(myset)
```

```python
# Remove Items
myset = {"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin"}
myset.remove("Audi")
print(myset)
```

```python
# Set Operations
A = {1,2,3,4}
B = {3,4,5,6}
print("Union : ",A | B)
print("Intersection : ",A & B)
print("Difference : ",A - B)
print("Symmetric Difference", A ^ B)
```

# Python FrozenSet  
frozenset is a immutable version of set.

```python
# frozenset
myset = frozenset({"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin"})
print(myset)
print(type(myset))
```

```python
# Adding or removal of elements is not allowed in frozenset
myset = frozenset({"Ferrari","Haas","McLaren","Mercedes","Audi","Aston Martin"})
myset.add("BMW")
myset.remove("Audi")
print(myset)

```

```python
# Merge frozensets using union
myset = frozenset({"Ferrari","McLaren","Mercedes","Aston Martin"})
newset = frozenset({"Apple","BlackBerry","Cherry"})
print("Union : ", myset | newset)
```

# Dictionary  
Dictionaries are used to store data values in key:value pairs  
A dictionary is a collection which is ordered, changeable and do not allow duplicates

```python
# Creation of Dictionary
cardata = {
    "brand":"Mercedes",
    "model":"AMG W17",
    "year":2026
}
print(cardata)
```

```python
# Retrieve Values from Dictionary
cardata = {
    "brand":"Mercedes",
    "model":"AMG W17",
    "year":2026
}
print(cardata['model'])
```

```python
# Dictionary takes the latest value given for a key. It will not allow duplicates.
cardata = {
    "brand":"Mercedes",
    "model":"AMG W17",
    "year":2026,
    "year":2017
}
print(cardata)
print("\n",cardata['year'])
```

```python
# Length of key-value pairs
len(cardata)
```

```python
# Changing/Adding key:value in dictionary
cardata = {
    "brand":"Mercedes",
    "model":"AMG W17",
    "year":2026
}
print(cardata)

cardata["color"] = 'red'
x = cardata.values()
print(x)
```

```python
'''3. A dictionary stores usernames and passwords. Write a python program to validate user login.'''
security = {
    "user1":"password1",
    "user2":"qwerty",
    "user3":"asdfg",
    "user4":"zxcv"
}
user = input("Enter username: ")
password = input("Enter password: ")

if security.get(user) == password:
    print("\nLogin Validation Successful")
else:
    print("\nLogin Validation Failed")
```

```python
'''4. A dictionary stores student names and marks. Write a program to find student who scored the highest marks.'''
marks = {
    "Lion":88,
    "Zebra":67,
    "Deer":39,
    "Fox":99
}
high = max(marks, key=marks.get)
print("Highest Scorer -")
print("Student Name: ",high)
print("Marks: ",marks[high])
```
