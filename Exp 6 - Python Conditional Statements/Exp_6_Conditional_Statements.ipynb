```python
#Salary Calculation with allowances

'''Problem Statement: Write a python program to calculate the gross salary of an employee based on basic salary.
If basic <= 10000: HRA = 20%, DA = 80%
If basic <= 20000: HRA = 25%, DA = 90%
Otherwise: HRA = 30%, DA = 95% '''

salary = int(input("Enter Basic Salary: "))

if salary <= 10000:
  hra = 0.2 * salary
  da = 0.8 * salary
  gross_salary = salary + hra + da
  print("Gross Salary: ",gross_salary)
elif salary <= 20000:
  hra = 0.25 * salary
  da = 0.9 * salary
  gross_salary = salary + hra + da
  print("Gross Salary: ",gross_salary)
else:
  hra = 0.3 * salary
  da = 0.95 * salary
  gross_salary = salary + hra + da
  print("Gross Salary: ",gross_salary)
```

```python
'''Problem Statement: Write a python to calculate income tax based on annual income:
Up to Rs. 2,50,000: No Tax
Rs. 2,50,001 - Rs. 5,00,000: 5%
Rs. 5,00,001 - Rs. 10,00,000: 20%
Above Rs. 10,00,000: 30% '''

income = int(input("Enter annual income: "))
tax = 0

if income <= 250000:
  tax = 0

elif income <= 500000:
  tax = (income - 250000) * 0.05

elif income <= 1000000:
  tax = (250000 * 0.05) + (income - 500000) * 0.20

else:
  tax = (250000 * 0.05) + (500000 * 0.20) + (income - 1000000) * 0.30

print("Income Tax to be paid: Rs.", tax)
print("In-Hand Salary: Rs. ",income-tax)
```
