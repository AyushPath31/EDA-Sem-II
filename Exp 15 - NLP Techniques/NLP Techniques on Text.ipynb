```python
from nltk.tokenize import word_tokenize

text = "Natural Language Processing is interesting"
tokens = word_tokenize(text)
print(tokens)
```

```python
# Sentence Tokenization

from nltk.tokenize import sent_tokenize

text = "NLP is useful. It is used in chatbots."
sentences = sent_tokenize(text)
print(sentences)
```

```python
# Stop Word Removal
# Stop words such as is, the, and, in are removed

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

words = word_tokenize(text)

filtered_words = [w for w in words if w.lower() not in stop_words]

print(filtered_words)
```

```python
# Stemming
# Stemming reduces words to their root form.

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["coding","designing","racing","running","flying","better"]

for w in words:
  print(stemmer.stem(w))
```

```python
# Lemmatization
# Lemmatization converts words to their dictionary root form

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["coding","designing","racing","running","flying","better"]

for w in words[:-1]:
  print(lemmatizer.lemmatize(w, pos='v')) # pos specifies type of word that is there. v is for verm, a is adjective

print()

print(lemmatizer.lemmatize("better", pos='a')) # better needs pos=a as it is adjective
```

```python
# Part-of-Speech (POS) Tagging
# POS tagging identifies grammatical roles of words.

import nltk
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag

words = word_tokenize("TypeScript is better than JavaScript")

pos_tags = pos_tag(words)
print(pos_tags)
```

### <u>Meaning of Tags </u>

```python
# Word Frquency count

from nltk.probability import FreqDist
text = "Ba Ba Black Sheep, have you any wool? yes sir, yes sir, three bags full"
words = word_tokenize(text)
fd = FreqDist(words)

for chars in fd:
  print(chars, fd[chars])

# or print(fd.most_common())
```
